## A* Algorithm Implementation

A* Algorithm

```python
import heapq

class Node:
    def __init__(self, name, x, y):
        self.name = name
        self.x = x
        self.y = y
        self.g_score = float('inf')
        self.f_score = float('inf')
        self.came_from = None

    def __lt__(self, other):
        return self.f_score < other.f_score

    def __repr__(self):
        return f"Node({self.name}, f={self.f_score:.2f}, g={self.g_score:.2f})"
```

```python
def heuristic(node1, node2):

    return ((node1.x - node2.x)**2 + (node1.y - node2.y)**2)**0.5

def reconstruct_path(current_node):
    path = []
    while current_node:
        path.append(current_node.name)
        current_node = current_node.came_from
    return path[::-1]

def a_star(graph, start_node, goal_node):
    start_node.g_score = 0
    start_node.f_score = heuristic(start_node, goal_node)

    open_set = [(start_node.f_score, start_node)]
    closed_set = set()

    while open_set:
        current_f_score, current_node = heapq.heappop(open_set)

        if current_node == goal_node:
            return reconstruct_path(current_node)

        if current_node in closed_set:
            continue

        closed_set.add(current_node)

        for neighbor_name, weight in graph[current_node.name].items():
            neighbor = next(n for n in all_nodes if n.name == neighbor_name)

            tentative_g_score = current_node.g_score + weight

            if tentative_g_score < neighbor.g_score:
                neighbor.came_from = current_node
                neighbor.g_score = tentative_g_score
                neighbor.f_score = neighbor.g_score + heuristic(neighbor, goal_node)
                heapq.heappush(open_set, (neighbor.f_score, neighbor))

    return None
```

```python

node_a = Node('A', 0, 0)
node_b = Node('B', 1, 3)
node_c = Node('C', 4, 1)
node_d = Node('D', 6, 4)
node_e = Node('E', 5, 0)
node_f = Node('F', 7, 2)

all_nodes = [node_a, node_b, node_c, node_d, node_e, node_f]

graph = {
    'A': {'B': 1, 'C': 4},
    'B': {'A': 1, 'D': 5, 'E': 2},
    'C': {'A': 4, 'F': 2},
    'D': {'B': 5, 'F': 1},
    'E': {'B': 2, 'F': 3},
    'F': {'C': 2, 'D': 1, 'E': 3}
}


node_map = {node.name: node for node in all_nodes}

start_node_name = 'A'
goal_node_name = 'F'

start_node = node_map[start_node_name]
goal_node = node_map[goal_node_name]

path = a_star(graph, start_node, goal_node)

if path:
    print(f"Shortest path from {start_node_name} to {goal_node_name}: {path}")
    print(f"Total cost: {goal_node.g_score}")
else:
    print(f"No path found from {start_node_name} to {goal_node_name}.")
```
