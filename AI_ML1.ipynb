## Q4 Detect and handle missing values using appropriate techniques such as deletion, mean, median, mode, or interpolation.

```python
# Handle missing values in 'Sale_Event' column
# For categorical columns with missing values, filling with a placeholder like 'No Sale Event' is a common approach.
df['Sale_Event'].fillna('No Sale Event', inplace=True)

print("Missing values after handling:")
display(df.isnull().sum())
```

## Q5 Identify duplicate records in the dataset and remove them.

```python
# Identify duplicate records
num_duplicates = df.duplicated().sum()
print(f"Number of duplicate records found: {num_duplicates}")

# Remove duplicate records
df.drop_duplicates(inplace=True)
print(f"Number of records after removing duplicates: {len(df)}")
```

## Q6 Detect outliers using statistical methods (IQR or Z-score) and visualize them using box plots.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Identify numerical columns for outlier detection
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns

print("Detecting outliers using IQR method and visualizing with box plots:")

plt.figure(figsize=(15, 10))
for i, col in enumerate(numerical_cols):
    plt.subplot(3, 3, i + 1) # Adjust subplot grid as needed
    sns.boxplot(x=df[col])
    plt.title(f'Box plot of {col}')

    # Calculate IQR and outlier boundaries
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    print(f"Column '{col}': {len(outliers)} outliers detected (IQR method)")

plt.tight_layout()
plt.show()
```

## Q7 Handle outliers using suitable preprocessing techniques.

```python


print("Handling outliers by capping at IQR bounds:")

for col in numerical_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df[col] = df[col].clip(lower=lower_bound, upper=upper_bound)
    print(f"Outliers in '{col}' capped.")

print("\nOutliers handling complete. Displaying descriptive statistics after capping:")
display(df.describe())
```

## Q8 Convert categorical variables into numerical form using: Label Encoding and One-Hot Encoding.

```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder

# Identify categorical columns (excluding 'Date' for now, as it will be handled in Q13)
categorical_cols = df.select_dtypes(include='object').columns.tolist()
categorical_cols.remove('Date') # Exclude 'Date' for now

print("Categorical columns identified:", categorical_cols)

# Check unique values for each categorical column to decide encoding strategy
print("\nUnique values per categorical column:")
for col in categorical_cols:
    print(f"  {col}: {df[col].nunique()} unique values")

# --- One-Hot Encoding ---
# Columns with a relatively low number of unique values are good candidates for One-Hot Encoding
one_hot_encode_cols = ['Platform', 'Product_Category', 'Condition', 'Sale_Event', 'Stock_Status']

print(f"\nApplying One-Hot Encoding to: {one_hot_encode_cols}")
df = pd.get_dummies(df, columns=one_hot_encode_cols, drop_first=True)


label_encode_cols = ['Model_Name']

print(f"Applying Label Encoding to: {label_encode_cols}")
le = LabelEncoder()
for col in label_encode_cols:
    df[col] = le.fit_transform(df[col])

print("\nDataset head after categorical encoding:")
display(df.head())
print("\nNew DataFrame info after encoding:")
df.info()
```

## Q12 Create new features from existing attributes through feature engineering.

```python
# Extract temporal features from 'Date'
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day
df['DayOfWeek'] = df['Date'].dt.dayofweek # Monday=0, Sunday=6

# Create price difference features
df['Price_Difference_USD'] = df['Launch_Price_USD'] - df['Current_Price_USD']
df['Price_Difference_INR'] = df['Launch_Price_INR'] - df['Current_Price_INR']

# Create price change percentage features (handle division by zero if Launch_Price_USD or INR can be 0)
df['Price_Change_Pct_USD'] = (df['Current_Price_USD'] - df['Launch_Price_USD']) / df['Launch_Price_USD'].replace(0, 1) * 100
df['Price_Change_Pct_INR'] = (df['Current_Price_INR'] - df['Launch_Price_INR']) / df['Launch_Price_INR'].replace(0, 1) * 100

# Assuming 'Launch_Price_USD' and 'Current_Price_USD' are directly related to the product value
# You could also create interaction terms or ratios, e.g., 'Rating_per_Review' or 'Price_per_Rating'
# For now, let's keep it simple with the above.

print("New features created. Displaying head of DataFrame with new features:")
display(df.head())

print("New DataFrame info after feature engineering:")
df.info()
```

## Q11 Generate and interpret a correlation matrix and heatmap for the dataset.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np # Import numpy

# Calculate the correlation matrix for numerical and encoded features
# Exclude 'Date' for correlation matrix as it's still an object type
correlation_matrix = df.drop(columns=['Date']).corr()

print("Correlation Matrix:")
display(correlation_matrix)

plt.figure(figsize=(15, 12))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix of Features')
plt.show()

# Interpretation and Feature Selection based on correlation
# Identify highly correlated features (e.g., absolute correlation > 0.9)
highly_correlated_pairs = []
corr_threshold = 0.9

# Get unique pairs for comparison
corr_upper_triangle = correlation_matrix.where(np.triu(np.ones(correlation_matrix.shape), k=1).astype(bool))

for column in corr_upper_triangle.columns:
    for row in corr_upper_triangle.index:
        if abs(corr_upper_triangle.loc[row, column]) > corr_threshold:
            highly_correlated_pairs.append((row, column, corr_upper_triangle.loc[row, column]))

if highly_correlated_pairs:
    print(f"\nHighly correlated feature pairs (absolute correlation > {corr_threshold}):")
    for f1, f2, corr_val in highly_correlated_pairs:
        print(f"  - {f1} and {f2}: {corr_val:.2f}")
    print("\nConsider removing one of the highly correlated features to reduce multicollinearity.")
    # For this example, let's decide to drop 'Launch_Price_INR' since 'Launch_Price_USD' is also present and highly correlated.
    # And 'Current_Price_INR' since 'Current_Price_USD' is present.
    # Note: The choice of which feature to drop can depend on domain knowledge or further analysis.

    columns_to_drop = []
    if ('Launch_Price_USD', 'Launch_Price_INR', correlation_matrix.loc['Launch_Price_USD', 'Launch_Price_INR']) in highly_correlated_pairs or \
       ('Launch_Price_INR', 'Launch_Price_USD', correlation_matrix.loc['Launch_Price_INR', 'Launch_Price_USD']) in highly_correlated_pairs:
        columns_to_drop.append('Launch_Price_INR')
    if ('Current_Price_USD', 'Current_Price_INR', correlation_matrix.loc['Current_Price_USD', 'Current_Price_INR']) in highly_correlated_pairs or \
       ('Current_Price_INR', 'Current_Price_USD', correlation_matrix.loc['Current_Price_INR', 'Current_Price_USD']) in highly_correlated_pairs:
        columns_to_drop.append('Current_Price_INR')

    if columns_to_drop:
        print(f"Dropping highly correlated columns: {columns_to_drop}")
        df.drop(columns=columns_to_drop, inplace=True)
        print("DataFrame after dropping highly correlated columns:")
        display(df.head())
        print("New DataFrame info after feature selection:")
        df.info()
else:
    print(f"\nNo highly correlated feature pairs (absolute correlation > {corr_threshold}) found.")
```

## Q9 Apply feature scaling techniques: Min-Max Normalization and Standardization (Z-score Scaling).

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# Identify numerical columns for scaling
# Exclude 'Date' for now, as it's not a numerical feature for direct scaling.
# 'Model_Name' has already been label encoded and can be scaled.
# Update columns_to_scale to reflect columns actually present in the DataFrame after previous preprocessing steps.
# Specifically, 'Launch_Price_INR' and 'Current_Price_INR' were dropped in Q10/Q11.

all_possible_numerical_cols = [
    'Launch_Price_USD', 'Launch_Price_INR', 'Current_Price_USD',
    'Current_Price_INR', 'Discount_Pct', 'Rating', 'Reviews_Count', 'Model_Name',
    'Price_Difference_USD', 'Price_Difference_INR', 'Price_Change_Pct_USD', 'Price_Change_Pct_INR'
]

# Filter the list to only include columns that are currently in the DataFrame
columns_to_scale = [col for col in all_possible_numerical_cols if col in df.columns]


df_scaled = df.copy()

print("Applying Min-Max Normalization...")
min_max_scaler = MinMaxScaler()
df_scaled[columns_to_scale] = min_max_scaler.fit_transform(df_scaled[columns_to_scale])
print("Min-Max Normalization applied to selected columns. Displaying head of scaled data:")
display(df_scaled[columns_to_scale].head())

# To demonstrate StandardScaler, we'll create another temporary DataFrame or overwrite,
# but for clarity, let's create a new set of columns if we want to keep both scaled versions.
# For this demonstration, we'll just show one type of scaling for the primary 'df' variable
# and print the head. If the user wants both permanently, we would create separate columns.

df_standard_scaled = df.copy()
print("\nApplying Standardization (Z-score Scaling)...")
standard_scaler = StandardScaler()
df_standard_scaled[columns_to_scale] = standard_scaler.fit_transform(df_standard_scaled[columns_to_scale])
print("Standardization applied to selected columns. Displaying head of standard scaled data:")
display(df_standard_scaled[columns_to_scale].head())

# For subsequent steps, let's assume we proceed with the Min-Max scaled DataFrame if not specified otherwise
# or apply scaling directly to `df` if only one is needed for the main workflow.
# For now, let's make the primary 'df' use Min-Max scaled values for numerical columns
# and keep the original 'Date' column as is for the next step (Q13).
# Note: If the intent is to use only one scaling method for further analysis, one would typically
# apply it directly to the main DataFrame or choose one of the created scaled DataFrames.

# For the purpose of moving forward with subsequent questions,
# let's replace the original numerical columns in 'df' with their Min-Max scaled versions.
df[columns_to_scale] = min_max_scaler.fit_transform(df[columns_to_scale])
print("\nNumerical columns in main DataFrame 'df' are now Min-Max scaled.")
display(df.head())
```
