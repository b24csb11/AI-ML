```python
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, cross_val_score, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                             roc_auc_score, ConfusionMatrixDisplay,
                             RocCurveDisplay, classification_report)
import joblib
import pandas as pd
df = pd.read_csv("student_pass_fail_strong.csv")
df.head()

print(df.shape)
print(df.isna().sum(), "\n")
print(df.describe().round(2), "\n")
print("Class balance:\n", df["passed"].value_counts(normalize=True).round(3))

df.drop(columns="student_id").hist(figsize=(10, 6), bins=15)
plt.tight_layout()
plt.show()
```

```python
df["hours_studied"]  = df["hours_studied"].where(df["hours_studied"].between(0, 12))
df["attendance_pct"] = df["attendance_pct"].where(df["attendance_pct"].between(0, 100))
df["previous_score"] = df["previous_score"].where(df["previous_score"].between(0, 100))
df["sleep_hours"]    = df["sleep_hours"].where(df["sleep_hours"].between(2, 12))

print(df.isna().sum())
```

```python
features = ["hours_studied", "attendance_pct", "previous_score", "sleep_hours"]
X = df[features]
y = df["passed"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=1, stratify=y)
print("Train:", X_train.shape, " Test:", X_test.shape)

model = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),   # fills missing values
    ("scaler", StandardScaler()),                    # mean 0, std 1
    ("logreg", LogisticRegression(max_iter=1000)),
])
```

```python
cv_acc = cross_val_score(model, X_train, y_train, cv=5, scoring="accuracy")
cv_auc = cross_val_score(model, X_train, y_train, cv=5, scoring="roc_auc")
print(f"CV accuracy: {cv_acc.mean():.3f} (+/- {cv_acc.std():.3f})")
print(f"CV ROC-AUC : {cv_auc.mean():.3f} (+/- {cv_auc.std():.3f})")

model.fit(X_train, y_train)
```

```python
param_grid = {
    "logreg__C": [0.1, 1, 10, 100],
    "logreg__class_weight": [None, "balanced"],
}
grid = GridSearchCV(model, param_grid, cv=5, scoring="balanced_accuracy")
grid.fit(X_train, y_train)

print("Best parameters:", grid.best_params_)
print(f"Best CV balanced accuracy: {grid.best_score_:.3f}")

best_model = grid.best_estimator_
```

```python
y_pred = best_model.predict(X_test)
y_prob = best_model.predict_proba(X_test)[:, 1]

print(f"Accuracy : {accuracy_score(y_test, y_pred):.3f}")
print(f"Precision: {precision_score(y_test, y_pred):.3f}")
print(f"Recall   : {recall_score(y_test, y_pred):.3f}")
print(f"F1 score : {f1_score(y_test, y_pred):.3f}")
print(f"ROC-AUC  : {roc_auc_score(y_test, y_prob):.3f}\n")
print(classification_report(y_test, y_pred,
                            target_names=["Fail", "Pass"], zero_division=0))

fig, ax = plt.subplots(1, 2, figsize=(11, 4))
ConfusionMatrixDisplay.from_estimator(best_model, X_test, y_test,
                                      display_labels=["Fail", "Pass"], ax=ax[0])
RocCurveDisplay.from_estimator(best_model, X_test, y_test, ax=ax[1])
ax[0].set_title("Confusion matrix")
ax[1].set_title("ROC curve")
plt.tight_layout()
plt.show()
```

```python
for t in [0.3, 0.4, 0.5, 0.6, 0.7]:
    pred_t = (y_prob >= t).astype(int)
    print(f"threshold={t:.1f}  accuracy={accuracy_score(y_test, pred_t):.3f}  "
          f"precision={precision_score(y_test, pred_t):.3f}  "
          f"recall={recall_score(y_test, pred_t):.3f}")

coefs = best_model.named_steps["logreg"].coef_[0]
importance = pd.Series(coefs, index=features).sort_values()
print("\n", importance.round(3))
importance.plot(kind="barh", title="Coefficients (scaled features)")
plt.show()
```

```python
import numpy as np

new_students = pd.DataFrame({
    "hours_studied":  [2.0, 5.5, 8.0],
    "attendance_pct": [60, 80, 95],
    "previous_score": [45, 60, 80],
    "sleep_hours":    [5.5, 7.0, 7.5],
})

result = new_students.copy()
result["pass_probability"] = best_model.predict_proba(new_students)[:, 1].round(3)
result["prediction"] = np.where(best_model.predict(new_students) == 1, "Pass", "Fail")
result
```
