## Uniform Cost Search (UCS)




```python
import heapq

def uniform_cost_search(graph, start, goal):


    priority_queue = [(0, start, [start])]

    min_cost_to_node = {start: 0}

    while priority_queue:
        current_cost, current_node, current_path = heapq.heappop(priority_queue)


        if current_cost > min_cost_to_node.get(current_node, float('inf')):
            continue

        if current_node == goal:
            return current_path, current_cost

        for neighbor, edge_cost in graph.get(current_node, {}).items():
            new_cost = current_cost + edge_cost


            if new_cost < min_cost_to_node.get(neighbor, float('inf')):
                min_cost_to_node[neighbor] = new_cost
                new_path = current_path + [neighbor]
                heapq.heappush(priority_queue, (new_cost, neighbor, new_path))

    return None, None
```

### Example Graph and Usage

Let's define a sample graph and use the `uniform_cost_search` function to find the lowest-cost path.

```python

example_graph = {
    'A': {'B': 1, 'C': 10},
    'B': {'D': 2},
    'C': {'D': 3},
    'D': {'E': 4, 'F': 1},
    'E': {'F': 5},
    'F': {}
}

start_node = 'A'
goal_node = 'F'

print(f"Searching for the lowest-cost path from {start_node} to {goal_node} in the graph:")
for node, neighbors in example_graph.items():
    print(f"  {node}: {neighbors}")
print("---")

path, cost = uniform_cost_search(example_graph, start_node, goal_node)

if path and cost is not None:
    print(f"Lowest-cost path from {start_node} to {goal_node}: {path}")
    print(f"Total cost: {cost}")
else:
    print(f"No path found from {start_node} to {goal_node}.")

print("\n--- Another Example ---")
start_node_2 = 'A'
goal_node_2 = 'E'
path_2, cost_2 = uniform_cost_search(example_graph, start_node_2, goal_node_2)

if path_2 and cost_2 is not None:
    print(f"Lowest-cost path from {start_node_2} to {goal_node_2}: {path_2}")
    print(f"Total cost: {cost_2}")
else:
    print(f"No path found from {start_node_2} to {goal_node_2}.")
```
